## 2.1. Using the Indexing Operator

The indexing operator `[]` is a fundamental way to access elements in both Pandas Series and DataFrames.

For **Series**, you can use `[]` with:
- A single label (index value) to get a single element.
- A list of labels to get multiple elements.
- A slice of labels to get a range of elements.
- A boolean array to filter elements.

For **DataFrames**, you can use `[]` with:
- A single column name (as a string) to select a Series (a single column).
- A list of column names to select a DataFrame (multiple columns).
- A boolean array to filter rows.


### **Task 15**. Use positional index to get all elements except the first one in the df **club_ranking**

### **Task 16**. Use label index to get all elements after MNC in the Series **club_ranking**

## 2.3. Using `.loc` and `.iloc`

To avoid confusion and ensure explicit indexing, Pandas provides two attribute accessors:

*   `.loc`: This is primarily used for **label-based indexing**. You use the actual labels (index names or column names) to select data.
*   `.iloc`: This is primarily used for **integer position-based indexing**. You use the numerical position of the rows and columns (starting from 0) to select data.

These accessors make your code more readable and predictable, especially when dealing with Series or DataFrames that have non-standard integer indexes.

### **Task 19**. use `.iloc` to get the second and third elements in the Series **colors**

In many data analysis tasks, you'll need to combine information from multiple sources. Pandas provides powerful tools to merge and concatenate DataFrames and Series.

The two most common methods for combining datasets are:

*   `.concat()`: Primarily used for stacking DataFrames or Series either vertically (adding rows) or horizontally (adding columns).
*   `.merge()`: Primarily used for combining DataFrames based on common columns or index values, similar to SQL joins.
